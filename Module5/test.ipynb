```python
import psycopg

#print('Connecting to the PostgreSQL database...')
conn = psycopg.connect(
    host="localhost",
    port='5432',
    dbname="postgres",
    user="postgres",
    password="postgres")
```

```python
cur = conn.cursor()
```

```python
# execute a statement
print('PostgreSQL database version:')
cur.execute('SELECT version()')

# display the PostgreSQL database server version
db_version = cur.fetchone()
print(db_version)
       
createCmd = """ CREATE TABLE parts (
                part_id SERIAL PRIMARY KEY,
                part_name VARCHAR(255) NOT NULL
                )
            """
    
cur.execute(createCmd)
conn.commit()
```

```python
cur.execute("INSERT into parts VALUES(1, 'wheels')")
cur.execute("INSERT into parts VALUES(2, 'door')")
cur.execute("INSERT into parts VALUES(3, 'seats')")
cur.execute("INSERT into parts VALUES(4, 'engine')")
conn.commit()
```

```python
queryCmd = 'SELECT * from parts'
cur.execute(queryCmd)    
print("The number of rows: ", cur.rowcount)

row = cur.fetchone()
while row is not None:
    print("part_id:", row[0], ", part_name", row[1])
    row = cur.fetchone()    
```

```python
part_id = 4
cur.execute("DELETE FROM parts WHERE part_id = %s", (part_id,))
conn.commit()
```

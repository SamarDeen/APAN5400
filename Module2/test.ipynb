Sets: Unique, Unordered

```python
cities_traveled = {"NYC", "London", "Edinburgh", "London", "New Orleans", "Boston", "Dubai", "Bangkok", "Cape Town"}
print(len(cities_traveled))
print(cities_traveled)
```

```python
cities_traveled.discard("London")
print(cities_traveled)
```

```python
USA = {"NYC", "New Orleans", "Boston"}
for cities in cities_traveled:
    if cities in USA:
        print(cities)
```

While loop and Break Statement

```python
n = 0

while n < len(cities_traveled):
    print(list(cities_traveled)[n])
    n += 1
```

Python Functions

```python
n = 11
if n <= 1:
    print(False)
else:
    prime = True

    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            prime = False
            break
    print(prime)
```

Now make this a reusable function

```python
def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

print(is_prime(12))

print(num) # num does not exist as it is a local variable
```
